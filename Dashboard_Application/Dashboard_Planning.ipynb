```python
# ============================================================
# DAY 1 – DASHBOARD PLANNING & APPLICATION SETUP
# ============================================================

import os
import pandas as pd

print("=" * 70)
print("DAY 1 – DASHBOARD PLANNING & APPLICATION SETUP")
print("=" * 70)

# ------------------------------------------------------------
# 1. Project Path
# ------------------------------------------------------------

project_root = r"D:\aiLAB99\Customer-Personality-Analysis"

print("\nProject Root:")
print(project_root)


# ------------------------------------------------------------
# 2. Review Important Project Outputs
# ------------------------------------------------------------

required_files = {

    "Raw Dataset":
        os.path.join(
            project_root,
            "01_Raw_Data",
            "marketing_campaign.csv"
        ),

    "Cleaned Dataset":
        os.path.join(
            project_root,
            "Cleaned_Data",
            "customer_personality_cleaned.csv"
        ),

    "Selected Features":
        os.path.join(
            project_root,
            "Customer_Segmentation",
            "Feature_Engineering",
            "Selected_Features",
            "clustering_selected_features.csv"
        ),

    "Feature Selection Report":
        os.path.join(
            project_root,
            "Customer_Segmentation",
            "Feature_Engineering",
            "Selected_Features",
            "clustering_feature_selection_report.csv"
        ),

    "Cluster Business Profile":
        os.path.join(
            project_root,
            "Customer_Segmentation",
            "Reports",
            "cluster_business_profile.csv"
        ),

    "Cluster Profile Comparison":
        os.path.join(
            project_root,
            "Customer_Segmentation",
            "Reports",
            "cluster_profile_comparison.csv"
        ),

    "Customer Cluster Profile":
        os.path.join(
            project_root,
            "Customer_Segmentation",
            "Reports",
            "customer_cluster_profile_summary.csv"
        ),

    "Final Algorithm Comparison":
        os.path.join(
            project_root,
            "Customer_Segmentation",
            "Reports",
            "final_algorithm_comparison.csv"
        ),

    "Cluster Radar Data":
        os.path.join(
            project_root,
            "Customer_Segmentation",
            "Reports",
            "cluster_radar_chart_data.csv"
        )
}


print("\n" + "=" * 70)
print("PROJECT OUTPUT CHECK")
print("=" * 70)


file_status = []

for name, path in required_files.items():

    exists = os.path.exists(path)

    file_status.append({
        "Output": name,
        "Status": "AVAILABLE" if exists else "NOT FOUND",
        "Path": path
    })

    print(
        f"{name}: "
        f"{'AVAILABLE' if exists else 'NOT FOUND'}"
    )


# ------------------------------------------------------------
# 3. Identify Final Customer Segmentation Dataset
# ------------------------------------------------------------

profiling_file = None

for root, dirs, files in os.walk(project_root):

    dirs[:] = [
        d for d in dirs
        if d not in ["venv", ".git", "__pycache__"]
    ]

    for file in files:

        if file.lower().endswith(".csv"):

            path = os.path.join(root, file)

            try:

                temp = pd.read_csv(
                    path,
                    nrows=5
                )

                if "KMeans_Cluster" in temp.columns:

                    profiling_file = path
                    break

            except Exception:
                continue

    if profiling_file:
        break


print("\n" + "=" * 70)
print("FINAL CUSTOMER SEGMENTATION DATASET")
print("=" * 70)

if profiling_file:

    print("FOUND:")
    print(os.path.abspath(profiling_file))

    final_df = pd.read_csv(
        profiling_file
    )

    print(
        "\nDataset Shape:",
        final_df.shape
    )

else:

    print(
        "No CSV containing KMeans_Cluster was found."
    )


# ------------------------------------------------------------
# 4. Identify Customer Segments
# ------------------------------------------------------------

if profiling_file:

    print("\n" + "=" * 70)
    print("CUSTOMER SEGMENT LABELS")
    print("=" * 70)

    cluster_summary = (
        final_df["KMeans_Cluster"]
        .value_counts()
        .sort_index()
        .reset_index()
    )

    cluster_summary.columns = [
        "Cluster",
        "Customers"
    ]

    cluster_summary["Percentage"] = (
        cluster_summary["Customers"]
        / len(final_df)
        * 100
    ).round(2)

    display(cluster_summary)


# ------------------------------------------------------------
# 5. Final Model Information
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL CUSTOMER SEGMENTATION MODEL")
print("=" * 70)

print("Algorithm: K-Means")
print("Number of Clusters (K): 2")
print("Model Status: FINAL")


# ------------------------------------------------------------
# 6. Final Engineered Features
# ------------------------------------------------------------

selected_features_file = required_files[
    "Selected Features"
]

print("\n" + "=" * 70)
print("FINAL ENGINEERED / SELECTED FEATURES")
print("=" * 70)

if os.path.exists(selected_features_file):

    selected_features_df = pd.read_csv(
        selected_features_file
    )

    print(
        "Selected Features File: AVAILABLE"
    )

    print(
        "Rows:",
        selected_features_df.shape[0]
    )

    print(
        "Columns:",
        selected_features_df.shape[1]
    )

    print("\nFeature File Preview:")

    display(
        selected_features_df.head()
    )

else:

    print(
        "Selected features file was not found."
    )


# ------------------------------------------------------------
# 7. Business Recommendations
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("BUSINESS RECOMMENDATIONS AVAILABLE")
print("=" * 70)

business_outputs = [
    "Customer Segment Summary",
    "Customer Personas",
    "Business Opportunities",
    "Marketing Strategy",
    "Product Recommendations",
    "Pricing and Discount Strategy",
    "Customer Retention Strategy",
    "Customer Reactivation Strategy",
    "Campaign Action Plan",
    "Marketing Strategy Matrix",
    "Business Dashboard",
    "Management Summary"
]

for item in business_outputs:

    print("✓", item)


# ------------------------------------------------------------
# 8. Dashboard Navigation Structure
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PROPOSED DASHBOARD NAVIGATION")
print("=" * 70)

navigation = [
    "1. Home / Project Overview",
    "2. Customer Segments",
    "3. Customer Demographics",
    "4. Spending & Purchasing Behavior",
    "5. Marketing & Campaign Analysis",
    "6. Business Recommendations",
    "7. Customer Personas",
    "8. About / Model Information"
]

for item in navigation:

    print(item)


# ------------------------------------------------------------
# 9. Dashboard Information
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DASHBOARD INFORMATION TO DISPLAY")
print("=" * 70)

dashboard_information = [
    "Number of Customers",
    "Number of Customer Segments",
    "Segment Distribution",
    "Average Income by Segment",
    "Average Spending by Segment",
    "Purchasing Behavior",
    "Shopping Channel Preferences",
    "Campaign Response",
    "Customer Recency",
    "Customer Risk / Activity",
    "Product Preferences",
    "Marketing Recommendations"
]

for item in dashboard_information:

    print("✓", item)


# ------------------------------------------------------------
# 10. Application Technology
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("APPLICATION TECHNOLOGY")
print("=" * 70)

print("Selected Technology: Streamlit")
print("Reason: Python-based and suitable for integrating ML results.")
print("Application Type: Interactive ML Business Dashboard")


# ------------------------------------------------------------
# 11. Save Day 1 Planning Report
# ------------------------------------------------------------

output_folder = os.path.join(
    project_root,
    "Customer_Insights",
    "Dashboard_Application"
)

os.makedirs(
    output_folder,
    exist_ok=True
)


planning_report = pd.DataFrame({

    "Item": [
        "Final Dataset",
        "Final Model",
        "Number of Clusters",
        "Dashboard Technology",
        "Customer Segmentation",
        "Business Recommendations",
        "Application Type"
    ],

    "Value": [
        os.path.basename(profiling_file)
        if profiling_file
        else "Not Found",

        "K-Means",

        "K = 2",

        "Streamlit",

        "Customer Personality Analysis",

        "Module 8 Business Insights",

        "Interactive ML Business Dashboard"
    ]
})


report_file = os.path.join(
    output_folder,
    "day_1_dashboard_planning_report.csv"
)

planning_report.to_csv(
    report_file,
    index=False
)


# ------------------------------------------------------------
# 12. Final Status
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DAY 1 FINAL STATUS")
print("=" * 70)

print(
    "Planning Report Saved:"
)

print(
    os.path.abspath(report_file)
)

print("\nDashboard Technology: Streamlit")

print("\nDAY 1 COMPLETED SUCCESSFULLY")
print("STATUS: PASS")
```
