# Task 5 – Website Engagement Analysis

This analysis examines customer interaction with the company's website. The analysis focuses on the number of monthly website visits, customer engagement levels, frequently visiting customers, and low-engagement customers.

Analyze Website Visits

```python
# Website visit statistics
print("Website Visit Statistics:")
print(df["NumWebVisitsMonth"].describe())
```

Count Website Visits

Create Website Visits Histogram

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="NumWebVisitsMonth",
    bins=20,
    kde=True
)

plt.title("Monthly Website Visits Distribution")
plt.xlabel("Number of Website Visits per Month")
plt.ylabel("Number of Customers")
plt.show()
```

Create Engagement Categories

```python
def classify_website_engagement(visits):
    if visits <= 3:
        return "Low Engagement"
    elif visits <= 7:
        return "Medium Engagement"
    else:
        return "High Engagement"

df["Website_Engagement"] = df["NumWebVisitsMonth"].apply(
    classify_website_engagement
)

print(df["Website_Engagement"].value_counts())
```

Engagement Summary

```python
engagement_counts = df["Website_Engagement"].value_counts()

print("Website Engagement Categories:")
print(engagement_counts)
```

```python
engagement_percentage = (
    df["Website_Engagement"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("Website Engagement Percentage:")
print(engagement_percentage)
```

Identify Frequently Visiting Customers

```python
frequent_visitors = df[
    df["NumWebVisitsMonth"] > 7
]

print("Frequently Visiting Customers:", len(frequent_visitors))
```

Identify Low-Engagement Customers

```python
low_engagement_customers = df[
    df["NumWebVisitsMonth"] <= 3
]

print("Low-Engagement Customers:", len(low_engagement_customers))
```

Engagement Count Plot

```python
plt.figure(figsize=(9, 6))

sns.countplot(
    data=df,
    x="Website_Engagement",
    order=["Low Engagement", "Medium Engagement", "High Engagement"]
)

plt.title("Customer Website Engagement Levels")
plt.xlabel("Website Engagement")
plt.ylabel("Number of Customers")
plt.show()
```

# Task 6 – Customer Recency Analysis

This analysis examines customer recency to understand how recently customers interacted with the company. The analysis identifies active, inactive, and at-risk customers and provides insights for customer re-engagement campaigns.

```python
print("Recency Statistics:")
print(df["Recency"].describe())
```

Recency Distribution

```python
recency_counts = df["Recency"].value_counts().sort_index()

print("Customer Count by Recency:")
print(recency_counts)
```

### Define Customer Activity Groups

For this analysis, we'll use:

Active: Recency ≤ 30 days
Moderately Active: 31–60 days
Inactive / At Risk: > 60 days

```python
def classify_recency(recency):
    if recency <= 30:
        return "Active"
    elif recency <= 60:
        return "Moderately Active"
    else:
        return "Inactive / At Risk"

df["Customer_Activity"] = df["Recency"].apply(classify_recency)

print("Customer Activity Categories:")
print(df["Customer_Activity"].value_counts())
```

Calculate Activity Percentages

```python
activity_percentage = (
    df["Customer_Activity"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("Customer Activity Percentage:")
print(activity_percentage)
```

Identify Active Customers

```python
active_customers = df[df["Recency"] <= 30]

print("Active Customers:", len(active_customers))
```

Identify Inactive / At-Risk Customers

```python
inactive_customers = df[df["Recency"] > 60]

print("Inactive / At-Risk Customers:", len(inactive_customers))
```

Identify Customers Needing Re-engagement

```python
reengagement_customers = df[df["Recency"] > 60]

print("Customers Requiring Re-engagement:", len(reengagement_customers))
```

Recency Histogram

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Recency",
    bins=20,
    kde=True
)

plt.title("Distribution of Customer Recency")
plt.xlabel("Recency (Days)")
plt.ylabel("Number of Customers")
plt.show()
```

Recency Box Plot

```python
plt.figure(figsize=(10, 5))

sns.boxplot(x=df["Recency"])

plt.title("Customer Recency Box Plot")
plt.xlabel("Recency (Days)")
plt.show()
```

Customer Activity Count Plot

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Customer_Activity",
    order=["Active", "Moderately Active", "Inactive / At Risk"]
)

plt.title("Customer Activity Levels")
plt.xlabel("Customer Activity")
plt.ylabel("Number of Customers")
plt.show()
```

Recency vs Total Spending

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

df["Total_Spending"] = df[["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds"]].sum(axis=1)

sns.scatterplot(
    data=df,
    x="Recency",
    y="Total_Spending"
)

plt.title("Customer Recency vs Total Spending")
plt.xlabel("Recency (Days)")
plt.ylabel("Total Spending")
plt.show()
```

Compare Activity With Spending

```python
activity_spending = df.groupby(
    "Customer_Activity"
)["Total_Spending"].agg(
    ["count", "mean", "sum"]
).round(2)

print("Spending by Customer Activity Level:")
print(activity_spending)
```

Find the Most Recent and Least Recent Customers

```python
print("Most Recently Active Customers:")
print(df[["ID", "Recency"]].sort_values("Recency").head(10))

print("\nLeast Recently Active Customers:")
print(df[["ID", "Recency"]].sort_values("Recency", ascending=False).head(10))
```
