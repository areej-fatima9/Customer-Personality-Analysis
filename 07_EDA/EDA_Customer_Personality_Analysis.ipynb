# Task 6 – Customer Recency Analysis

This analysis examines customer recency to understand how recently customers interacted with the company. The analysis identifies active, inactive, and at-risk customers and provides insights for customer re-engagement campaigns.

```python
print("Recency Statistics:")
print(df["Recency"].describe())
```

Recency Distribution

```python
recency_counts = df["Recency"].value_counts().sort_index()

print("Customer Count by Recency:")
print(recency_counts)
```

### Define Customer Activity Groups

For this analysis, we'll use:

Active: Recency ≤ 30 days
Moderately Active: 31–60 days
Inactive / At Risk: > 60 days

```python
def classify_recency(recency):
    if recency <= 30:
        return "Active"
    elif recency <= 60:
        return "Moderately Active"
    else:
        return "Inactive / At Risk"

df["Customer_Activity"] = df["Recency"].apply(classify_recency)

print("Customer Activity Categories:")
print(df["Customer_Activity"].value_counts())
```

Calculate Activity Percentages

```python
activity_percentage = (
    df["Customer_Activity"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("Customer Activity Percentage:")
print(activity_percentage)
```

Identify Active Customers

```python
active_customers = df[df["Recency"] <= 30]

print("Active Customers:", len(active_customers))
```

Identify Inactive / At-Risk Customers

```python
inactive_customers = df[df["Recency"] > 60]

print("Inactive / At-Risk Customers:", len(inactive_customers))
```

Identify Customers Needing Re-engagement

```python
reengagement_customers = df[df["Recency"] > 60]

print("Customers Requiring Re-engagement:", len(reengagement_customers))
```

Recency Histogram

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Recency",
    bins=20,
    kde=True
)

plt.title("Distribution of Customer Recency")
plt.xlabel("Recency (Days)")
plt.ylabel("Number of Customers")
plt.show()
```

Recency Box Plot

```python
plt.figure(figsize=(10, 5))

sns.boxplot(x=df["Recency"])

plt.title("Customer Recency Box Plot")
plt.xlabel("Recency (Days)")
plt.show()
```

Customer Activity Count Plot

Recency vs Total Spending

```python
complaint_income = df.groupby(
    "Complain"
)["Income"].agg(
    ["count", "mean", "median", "sum"]
).round(2)

print("Income Comparison – Complaints vs No Complaints:")
print(complaint_income)
```

Complaint Rate by Income Group

# Task 9 – Correlation Analysis

This analysis investigates relationships among numerical variables in the Customer Personality Analysis dataset. Pearson correlation is used to measure the strength and direction of relationships between variables. A correlation heatmap is also created to identify strong positive, strong negative, and weak relationships.

Select Numerical Variables

```python
numerical_columns = [
    "Income",
    "Kidhome",
    "Teenhome",
    "Recency",
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds",
    "NumDealsPurchases",
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases",
    "NumWebVisitsMonth",
    "AcceptedCmp1",
    "AcceptedCmp2",
    "AcceptedCmp3",
    "AcceptedCmp4",
    "AcceptedCmp5",
    "Complain",
    "Response",
    "Total_Spending"
]

print("Number of numerical variables:", len(numerical_columns))
print(numerical_columns)
```

Create Correlation Heatmap

```python
plt.figure(figsize=(16, 12))

correlation_matrix = df[numerical_columns].corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Pearson Correlation Heatmap")
plt.tight_layout()
plt.show()
```

Find Strong Positive Correlations

```python
strong_positive = []

for i in range(len(correlation_matrix.columns)):
    for j in range(i + 1, len(correlation_matrix.columns)):
        correlation = correlation_matrix.iloc[i, j]

        if correlation >= 0.50:
            strong_positive.append({
                "Variable 1": correlation_matrix.columns[i],
                "Variable 2": correlation_matrix.columns[j],
                "Correlation": round(correlation, 3)
            })

strong_positive_df = pd.DataFrame(strong_positive)

print("Strong Positive Correlations (r >= 0.50):")
print(strong_positive_df)
```

Find Strong Negative Correlations

```python
strong_negative = []

for i in range(len(correlation_matrix.columns)):
    for j in range(i + 1, len(correlation_matrix.columns)):
        correlation = correlation_matrix.iloc[i, j]

        if correlation <= -0.50:
            strong_negative.append({
                "Variable 1": correlation_matrix.columns[i],
                "Variable 2": correlation_matrix.columns[j],
                "Correlation": round(correlation, 3)
            })

strong_negative_df = pd.DataFrame(strong_negative)

print("Strong Negative Correlations (r <= -0.50):")
print(strong_negative_df)
```

### Find Weak Relationships
We'll identify pairs where the absolute correlation is below 0.10.

```python
weak_correlations = []

for i in range(len(correlation_matrix.columns)):
    for j in range(i + 1, len(correlation_matrix.columns)):
        correlation = correlation_matrix.iloc[i, j]

        if abs(correlation) < 0.10:
            weak_correlations.append({
                "Variable 1": correlation_matrix.columns[i],
                "Variable 2": correlation_matrix.columns[j],
                "Correlation": round(correlation, 3)
            })

weak_correlations_df = pd.DataFrame(weak_correlations)

print("Weak Correlations (|r| < 0.10):")
print(weak_correlations_df)
```

Find the Strongest Relationships

```python
correlation_pairs = []

for i in range(len(correlation_matrix.columns)):
    for j in range(i + 1, len(correlation_matrix.columns)):
        correlation_pairs.append({
            "Variable 1": correlation_matrix.columns[i],
            "Variable 2": correlation_matrix.columns[j],
            "Correlation": correlation_matrix.iloc[i, j]
        })

correlation_pairs_df = pd.DataFrame(correlation_pairs)

correlation_pairs_df["Absolute Correlation"] = (
    correlation_pairs_df["Correlation"].abs()
)

strongest_relationships = correlation_pairs_df.sort_values(
    "Absolute Correlation",
    ascending=False
).head(15)

print("Top 15 Strongest Correlations:")
print(
    strongest_relationships[
        ["Variable 1", "Variable 2", "Correlation"]
    ].round(3)
)
```

# Task 10 – Customer Segmentation

Customer segmentation is performed to identify groups of customers with similar purchasing behavior, spending patterns, campaign responses, and activity levels. The segments can help the business develop targeted marketing strategies.

Check Required Columns

```python
print("Required columns:")
print([
    "Income",
    "Total_Spending",
    "Total_Purchases",
    "NumDealsPurchases",
    "Total_Campaign_Acceptances",
    "Response",
    "Recency"
])
```

### Create High-Value and Low-Value Segments

We'll use Total Spending to identify customer value.

We'll define:

Top 25% → High-Value

Bottom 25% → Low-Value

Middle 50% → Medium-Value

```python
q25 = df["Total_Spending"].quantile(0.25)
q75 = df["Total_Spending"].quantile(0.75)

def classify_customer_value(spending):
    if spending >= q75:
        return "High-Value"
    elif spending <= q25:
        return "Low-Value"
    else:
        return "Medium-Value"

df["Customer_Value"] = df["Total_Spending"].apply(
    classify_customer_value
)

print("Customer Value Segments:")
print(df["Customer_Value"].value_counts())

print("\nValue Segment Percentages:")
print(
    df["Customer_Value"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

Identify Frequent Buyers

```python
df["Total_Purchases"] = df[
    ["NumDealsPurchases", "NumWebPurchases", "NumCatalogPurchases", "NumStorePurchases"]
].sum(axis=1)

purchase_threshold = df["Total_Purchases"].quantile(0.75)

df["Buyer_Frequency"] = df["Total_Purchases"].apply(
    lambda x: "Frequent Buyer"
    if x >= purchase_threshold
    else "Regular/Low-Frequency Buyer"
)

print("Buyer Frequency Segments:")
print(df["Buyer_Frequency"].value_counts())

print("\nPurchase Threshold:", purchase_threshold)

df["Buyer_Frequency"] = df["Total_Purchases"].apply(
    lambda x: "Frequent Buyer"
    if x >= purchase_threshold
    else "Regular/Low-Frequency Buyer"
)

print("Buyer Frequency Segments:")
print(df["Buyer_Frequency"].value_counts())

print("\nPurchase Threshold:", purchase_threshold)
```

Identify Discount Seekers

```python
deal_threshold = df["NumDealsPurchases"].quantile(0.75)

df["Discount_Behavior"] = df["NumDealsPurchases"].apply(
    lambda x: "Discount Seeker"
    if x >= deal_threshold
    else "Regular Buyer"
)

print("Discount Behavior:")
print(df["Discount_Behavior"].value_counts())

print("\nDiscount Purchase Threshold:", deal_threshold)
```

Identify Campaign Responders

```python
df["Campaign_Response_Group"] = df["Total_Campaign_Acceptances"].apply(
    lambda x: "Campaign Responder"
    if x > 0
    else "Non-Responder"
)

print("Campaign Response Segments:")
print(df["Campaign_Response_Group"].value_counts())
```

Identify Loyal Customers

```python
spending_threshold = df["Total_Spending"].quantile(0.75)
purchase_threshold = df["Total_Purchases"].quantile(0.75)

df["Loyalty_Segment"] = np.where(
    (df["Total_Spending"] >= spending_threshold) &
    (df["Total_Purchases"] >= purchase_threshold),
    "Loyal Customer",
    "Other Customer"
)

print("Loyalty Segments:")
print(df["Loyalty_Segment"].value_counts())
```

Identify Inactive Customers
