# Task 5: Identify Missing Values

## Objective
The objective of this task is to identify missing values in the dataset and calculate the percentage of missing values for each variable.

```python
# Missing values in each column
missing_values = df.isnull().sum()

missing_values
```

```python
# Percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage
```

```python
# Create a missing values summary table
missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_report
```

```python
# Columns without missing values
no_missing = df.columns[df.isnull().sum() == 0]

print(no_missing)
```

# Task 7: Identify Incorrect or Inconsistent Values

## Objective

The objective of this task is to identify incorrect, invalid, or inconsistent values in the Customer Personality Analysis dataset before preprocessing.

### Check Invalid Ages


```python
from datetime import datetime

current_year = datetime.now().year

df["Age"] = current_year - df["Year_Birth"]

df["Age"].describe()
```

### check unrealistic ages

```python
invalid_age = df[(df["Age"] < 18) | (df["Age"] > 100)]

print("Invalid Age Records:", len(invalid_age))

invalid_age
```

### Check Negative Income

```python
negative_income = df[df["Income"] < 0]

print("Negative Income Records:", len(negative_income))

negative_income
```

### Check Empty Values in Categorical Columns

```python
categorical_columns = ["Education", "Marital_Status"]

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts(dropna=False))
```

### Check Date Format

```python
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    format="%d-%m-%Y",
    errors="coerce"
)

print("Invalid Dates:", df["Dt_Customer"].isnull().sum())
```

# Task 9: Review Categorical Variables

## Objective

The objective of this task is to identify all categorical variables in the dataset and examine their categories before preprocessing.

```python
# Identify categorical columns
categorical_columns = df.select_dtypes(include=["object"]).columns

print("Categorical Variables:")
print(categorical_columns)
```

```python
# Number of categories and category names
for column in categorical_columns:
    print("=" * 50)
    print(f"Variable: {column}")
    print(f"Number of Categories: {df[column].nunique()}")
    print("Categories:")
    print(df[column].unique())
    print()
```

```python
categorical_summary = pd.DataFrame({
    "Variable": categorical_columns,
    "Number of Categories": [df[col].nunique() for col in categorical_columns],
    "Categories": [", ".join(map(str, df[col].unique())) for col in categorical_columns]
})

categorical_summary
```

# Task 10: Business Understanding

## Objective

The objective of this task is to identify important business indicators in the Customer Personality Analysis dataset and understand how they can support marketing and customer relationship decisions.

```python
# Business Indicators

business_indicators = {
    "High Income Customers": (df["Income"] > df["Income"].median()).sum(),
    "Frequent Buyers": (df["NumStorePurchases"] > df["NumStorePurchases"].median()).sum(),
    "Campaign Responders": df["Response"].sum(),
    "Customers with Complaints": df["Complain"].sum(),
    "High Website Visitors": (df["NumWebVisitsMonth"] > df["NumWebVisitsMonth"].median()).sum(),
    "High Spending Customers": (
        df["MntWines"] +
        df["MntFruits"] +
        df["MntMeatProducts"] +
        df["MntFishProducts"] +
        df["MntSweetProducts"] +
        df["MntGoldProds"]
    ).gt(
        (
            df["MntWines"] +
            df["MntFruits"] +
            df["MntMeatProducts"] +
            df["MntFishProducts"] +
            df["MntSweetProducts"] +
            df["MntGoldProds"]
        ).median()
    ).sum(),
    "Recently Active Customers": (df["Recency"] < 30).sum()
}

business_indicators
```

```python
business_summary = pd.DataFrame(
    business_indicators.items(),
    columns=["Business Indicator", "Count"]
)

business_summary
```
