# Activity 9 – Detect Income and Spending Outliers

## Objective

The objective of this activity is to identify outliers in numerical variables using the Interquartile Range (IQR) method and visualize them using boxplots. Outlier detection helps identify extreme values that may affect statistical analysis and machine learning models.

```python
import matplotlib.pyplot as plt
```

Select Numerical Variables

```python
columns = [
    "Income",
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]
```

Detect Outliers Using IQR

```python
outlier_summary = []

for col in columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    outlier_summary.append({
        "Variable": col,
        "Outliers": len(outliers),
        "Lower Bound": round(lower,2),
        "Upper Bound": round(upper,2)
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df
```

Create Boxplots

```python
for col in columns:

    plt.figure(figsize=(6,4))
    plt.boxplot(df[col], vert=False)

    plt.title(f"Boxplot of {col}")

    plt.xlabel(col)

    plt.show()
```

# Activity 10 – Handle Outliers

## Objective

The objective of this activity is to handle the outliers identified in the numerical variables. Instead of removing valuable customer records, outliers will be capped using the Interquartile Range (IQR) method to preserve the dataset while reducing the influence of extreme values.

Apply Winsorization (Capping)

```python
# Columns to cap
columns = [
    "Income",
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]

for col in columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Cap values
    df[col] = df[col].clip(lower=lower, upper=upper)

print("Outliers have been capped successfully.")
```

Verify the Results

```python
# Verify remaining outliers
summary = []

for col in columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    remaining = df[(df[col] < lower) | (df[col] > upper)]

    summary.append({
        "Variable": col,
        "Remaining Outliers": len(remaining)
    })

verification = pd.DataFrame(summary)

verification
```

Draw Boxplots Again

```python
import matplotlib.pyplot as plt

for col in columns:

    plt.figure(figsize=(6,4))

    plt.boxplot(df[col], vert=False)

    plt.title(f"After Outlier Handling - {col}")

    plt.xlabel(col)

    plt.show()
```

# Activity 14 – Before-and-After Comparison Report

```python
import pandas as pd

# Load original dataset
before = pd.read_csv("../01_Raw_Data/marketing_campaign.csv", sep="\t")

# Load cleaned dataset
after = pd.read_csv("../Cleaned_Data/customer_personality_cleaned.csv")
```

```python
# Create Age columns
before["Age"] = 2025 - before["Year_Birth"]

if "Age" not in after.columns:
    after["Age"] = 2025 - after["Year_Birth"]

# Function to count outliers using IQR
def count_outliers(df, cols):
    total = 0
    for col in cols:
        if col in df.columns:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1

            lower = Q1 - 1.5 * IQR
            upper = Q3 + 1.5 * IQR

            total += ((df[col] < lower) | (df[col] > upper)).sum()

    return total


numeric_cols = [
    "Income",
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]

comparison = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Missing Values",
        "Duplicate Rows",
        "Duplicate Customer IDs",
        "Invalid Dates",
        "Unrealistic Ages (>100)",
        "Outliers"
    ],

    "Before Cleaning": [
        len(before),
        before.shape[1],
        before.isnull().sum().sum(),
        before.duplicated().sum(),
        before["ID"].duplicated().sum(),
        pd.to_datetime(
            before["Dt_Customer"],
            format="%d-%m-%Y",
            errors="coerce"
        ).isnull().sum(),
        (before["Age"] > 100).sum(),
        count_outliers(before, numeric_cols)
    ],

    "After Cleaning": [
        len(after),
        after.shape[1],
        after.isnull().sum().sum(),
        after.duplicated().sum(),
        after["ID"].duplicated().sum(),
        pd.to_datetime(
            after["Dt_Customer"],
            errors="coerce"
        ).isnull().sum(),
        (after["Age"] > 100).sum(),
        count_outliers(after, numeric_cols)
    ]
})

comparison
```
